```python
import pandas as pd 
import matplotlib.pyplot as plt

#df = pd.read_csv('../dataset/final_dataset_diff_f_L10.csv')
df = pd.read_csv('../dataset/KNN_predict.csv')
df
```

```python


# Supposons que la colonne des dates s'appelle 'date' et est au format 'YYYY-MM-DD'
df['GAME_DATE'] = pd.to_datetime(df['GAME_DATE'])

# Extraire le mois et l'année des dates
df['month'] = df['GAME_DATE'].dt.to_period('M')

# Compter le nombre de matchs par mois
matches_per_month = df[df['HOME_WON'] == df['PRED']].groupby('month').size() / df.groupby('month').size() * 100

# Préparer les données pour le tracé
months = matches_per_month.index.astype(str)
match_counts = matches_per_month.values

# Tracer le graphique
plt.figure(figsize=(12, 6))
plt.bar(months, match_counts, color='skyblue')
plt.xlabel('Mois')
plt.ylabel('Pourcentage de prédiction')
plt.title('Pourcentage de bonne prédiction par mois')
plt.xticks(rotation=45)
plt.tight_layout()

# Afficher le graphique
plt.show()
```
