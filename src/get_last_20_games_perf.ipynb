```python
import pandas as pd 
```

```python
data_type = {"gameId": str, "teamId": str}
df_stat = pd.read_csv('../dataset/merged_team_stats.csv', dtype=data_type )
df_stat
```

```python
df_stat_last_20_perf = pd.DataFrame(columns=['gameId', 'teamId', 'GAME_DATE', 'DUEL','WL','NB_WIN_L20', 'PCT_TIR_REUSSI_L20','PCT_3PTS_L20', 
                           'PCT_LANCER_FRANC_L20','estimatedOffensiveRating_L20', 'offensiveRating_L20',
                           'estimatedDefensiveRating_L20',
                           'defensiveRating_L20','estimatedNetRating_L20', 'netRating_L20', 'assistPercentage_L20',
                           'assistToTurnover_L20', 'assistRatio_L20','estimatedTeamTurnoverPercentage_L20', 'turnoverRatio_L20',
                           'effectiveFieldGoalPercentage_L20','trueShootingPercentage_L20', 'estimatedPace_L20', 'pace_L20',
                           'pacePer40_L20', 'PIE_L20'])

for team in list(set(df_stat['teamId'].tolist())):
    team_stats = df_stat[df_stat['teamId'] == team]
    team_stats.reset_index(drop=True, inplace=True)
    index = 0
    while index < len(team_stats):
        df_last_20_games = team_stats.iloc[max(0, index - 20):index]
        df_last_20_games.reset_index(drop=True, inplace=True)
        new_row = {"gameId" :  team_stats['gameId'][index],
                   "teamId" : team_stats['teamId'][index],
                   "GAME_DATE" : team_stats['GAME_DATE'][index],
                   "DUEL" : team_stats['DUEL'][index],
                   "WL" : team_stats['WL'][index],
                   "NB_WIN_L20" : (df_last_20_games['WL'] == 'W').sum(),
                   "PCT_TIR_REUSSI_L20" : df_last_20_games['PCT_TIR_REUSSI'].mean(),
                   "PCT_3PTS_L20" : df_last_20_games['PCT_3PTS'].mean(),
                   "PCT_LANCER_FRANC_L20" : df_last_20_games['PCT_LANCER_FRANC'].mean(),
                   "estimatedOffensiveRating_L20" : df_last_20_games['estimatedOffensiveRating'].mean(),
                   "offensiveRating_L20" : df_last_20_games['offensiveRating'].mean(),
                   "estimatedDefensiveRating_L20" : df_last_20_games['estimatedDefensiveRating'].mean(),
                   "defensiveRating_L20" : df_last_20_games['defensiveRating'].mean(),
                   "estimatedNetRating_L20" : df_last_20_games['estimatedNetRating'].mean(),
                   "netRating_L20" : df_last_20_games['netRating'].mean(),
                   "assistPercentage_L20" : df_last_20_games['assistPercentage'].mean(),
                   "assistToTurnover_L20" : df_last_20_games['assistToTurnover'].mean(),
                   "assistRatio_L20" : df_last_20_games['assistRatio'].mean(),
                   "estimatedTeamTurnoverPercentage_L20" : df_last_20_games['estimatedTeamTurnoverPercentage'].mean(),
                   "turnoverRatio_L20" : df_last_20_games['turnoverRatio'].mean(),
                   "effectiveFieldGoalPercentage_L20" : df_last_20_games['effectiveFieldGoalPercentage'].mean(),
                   "trueShootingPercentage_L20" : df_last_20_games['trueShootingPercentage'].mean(),
                   "estimatedPace_L20" : df_last_20_games['estimatedPace'].mean(),
                   "pace_L20" : df_last_20_games['pace'].mean(),
                   "pacePer40_L20" : df_last_20_games['pacePer40'].mean(),
                   "PIE_L20" : df_last_20_games['PIE'].mean()}
        df_stat_last_20_perf.loc[len(df_stat_last_20_perf)] = new_row
        index += 1
df_stat_last_20_perf
    
```

```python
df_stat_last_20_perf.to_csv('../dataset/team_stat_last_20.csv', index=False)
```
