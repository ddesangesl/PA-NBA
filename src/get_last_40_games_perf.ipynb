```python
import pandas as pd 
```

```python
data_type = {"gameId": str, "teamId": str}
df_stat = pd.read_csv('../dataset/merged_team_stats.csv', dtype=data_type )
df_stat
```

```python
df_stat_last_40_perf = pd.DataFrame(columns=['gameId', 'teamId', 'GAME_DATE', 'DUEL','WL','NB_WIN_L40', 'PCT_TIR_REUSSI_L40','PCT_3PTS_L40', 
                           'PCT_LANCER_FRANC_L40','estimatedOffensiveRating_L40', 'offensiveRating_L40',
                           'estimatedDefensiveRating_L40',
                           'defensiveRating_L40','estimatedNetRating_L40', 'netRating_L40', 'assistPercentage_L40',
                           'assistToTurnover_L40', 'assistRatio_L40','estimatedTeamTurnoverPercentage_L40', 'turnoverRatio_L40',
                           'effectiveFieldGoalPercentage_L40','trueShootingPercentage_L40', 'estimatedPace_L40', 'pace_L40',
                           'pacePer40_L40', 'PIE_L40'])

for team in list(set(df_stat['teamId'].tolist())):
    team_stats = df_stat[df_stat['teamId'] == team]
    team_stats.reset_index(drop=True, inplace=True)
    index = 0
    while index < len(team_stats):
        df_last_40_games = team_stats.iloc[max(0, index - 40):index]
        df_last_40_games.reset_index(drop=True, inplace=True)
        new_row = {"gameId" :  team_stats['gameId'][index],
                   "teamId" : team_stats['teamId'][index],
                   "GAME_DATE" : team_stats['GAME_DATE'][index],
                   "DUEL" : team_stats['DUEL'][index],
                   "WL" : team_stats['WL'][index],
                   "NB_WIN_L40" : (df_last_40_games['WL'] == 'W').sum(),
                   "PCT_TIR_REUSSI_L40" : df_last_40_games['PCT_TIR_REUSSI'].mean(),
                   "PCT_3PTS_L40" : df_last_40_games['PCT_3PTS'].mean(),
                   "PCT_LANCER_FRANC_L40" : df_last_40_games['PCT_LANCER_FRANC'].mean(),
                   "estimatedOffensiveRating_L40" : df_last_40_games['estimatedOffensiveRating'].mean(),
                   "offensiveRating_L40" : df_last_40_games['offensiveRating'].mean(),
                   "estimatedDefensiveRating_L40" : df_last_40_games['estimatedDefensiveRating'].mean(),
                   "defensiveRating_L40" : df_last_40_games['defensiveRating'].mean(),
                   "estimatedNetRating_L40" : df_last_40_games['estimatedNetRating'].mean(),
                   "netRating_L40" : df_last_40_games['netRating'].mean(),
                   "assistPercentage_L40" : df_last_40_games['assistPercentage'].mean(),
                   "assistToTurnover_L40" : df_last_40_games['assistToTurnover'].mean(),
                   "assistRatio_L40" : df_last_40_games['assistRatio'].mean(),
                   "estimatedTeamTurnoverPercentage_L40" : df_last_40_games['estimatedTeamTurnoverPercentage'].mean(),
                   "turnoverRatio_L40" : df_last_40_games['turnoverRatio'].mean(),
                   "effectiveFieldGoalPercentage_L40" : df_last_40_games['effectiveFieldGoalPercentage'].mean(),
                   "trueShootingPercentage_L40" : df_last_40_games['trueShootingPercentage'].mean(),
                   "estimatedPace_L40" : df_last_40_games['estimatedPace'].mean(),
                   "pace_L40" : df_last_40_games['pace'].mean(),
                   "pacePer40_L40" : df_last_40_games['pacePer40'].mean(),
                   "PIE_L40" : df_last_40_games['PIE'].mean()}
        df_stat_last_40_perf.loc[len(df_stat_last_40_perf)] = new_row
        index += 1
df_stat_last_40_perf
    
```

```python
df_stat_last_40_perf.to_csv('../dataset/team_stat_last_40.csv', index=False)
```
