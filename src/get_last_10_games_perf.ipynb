```python
df_stat_last_10_perf = pd.DataFrame(columns=['gameId', 'teamId', 'GAME_DATE', 'DUEL','WL','NB_WIN', 'PCT_TIR_REUSSI_L10','PCT_3PTS_L10', 
                           'PCT_LANCER_FRANC_L10','estimatedOffensiveRating_L10', 'offensiveRating_L10',
                           'estimatedDefensiveRating_L10',
                           'defensiveRating_L10','estimatedNetRating_L10', 'netRating_L10', 'assistPercentage_L10',
                           'assistToTurnover_L10', 'assistRatio_L10','estimatedTeamTurnoverPercentage_L10', 'turnoverRatio_L10',
                           'effectiveFieldGoalPercentage_L10','trueShootingPercentage_L10', 'estimatedPace_L10', 'pace_L10',
                           'pacePer40_L10', 'PIE_L10'])

for team in list(set(df_stat['teamId'].tolist())):
    team_stats = df_stat[df_stat['teamId'] == team]
    team_stats.reset_index(drop=True, inplace=True)
    index = 0
    while index < len(team_stats):
        df_last_10_games = team_stats.iloc[max(0, index - 10):index]
        df_last_10_games.reset_index(drop=True, inplace=True)
        new_row = {"gameId" :  team_stats['gameId'][index],
                   "teamId" : team_stats['teamId'][index],
                   "GAME_DATE" : team_stats['GAME_DATE'][index],
                   "DUEL" : team_stats['DUEL'][index],
                   "WL" : team_stats['WL'][index],
                   "NB_WIN" : (df_last_10_games['WL'] == 'W').sum(),
                   "PCT_TIR_REUSSI_L10" : df_last_10_games['PCT_TIR_REUSSI'].mean(),
                   "PCT_3PTS_L10" : df_last_10_games['PCT_3PTS'].mean(),
                   "PCT_LANCER_FRANC_L10" : df_last_10_games['PCT_LANCER_FRANC'].mean(),
                   "estimatedOffensiveRating_L10" : df_last_10_games['estimatedOffensiveRating'].mean(),
                   "offensiveRating_L10" : df_last_10_games['offensiveRating'].mean(),
                   "estimatedDefensiveRating_L10" : df_last_10_games['estimatedDefensiveRating'].mean(),
                   "defensiveRating_L10" : df_last_10_games['defensiveRating'].mean(),
                   "estimatedNetRating_L10" : df_last_10_games['estimatedNetRating'].mean(),
                   "netRating_L10" : df_last_10_games['netRating'].mean(),
                   "assistPercentage_L10" : df_last_10_games['assistPercentage'].mean(),
                   "assistToTurnover_L10" : df_last_10_games['assistToTurnover'].mean(),
                   "assistRatio_L10" : df_last_10_games['assistRatio'].mean(),
                   "estimatedTeamTurnoverPercentage_L10" : df_last_10_games['estimatedTeamTurnoverPercentage'].mean(),
                   "turnoverRatio_L10" : df_last_10_games['turnoverRatio'].mean(),
                   "effectiveFieldGoalPercentage_L10" : df_last_10_games['effectiveFieldGoalPercentage'].mean(),
                   "trueShootingPercentage_L10" : df_last_10_games['trueShootingPercentage'].mean(),
                   "estimatedPace_L10" : df_last_10_games['estimatedPace'].mean(),
                   "pace_L10" : df_last_10_games['pace'].mean(),
                   "pacePer40_L10" : df_last_10_games['pacePer40'].mean(),
                   "PIE_L10" : df_last_10_games['PIE'].mean()}
        df_stat_last_10_perf.loc[len(df_stat_last_10_perf)] = new_row
        index += 1
df_stat_last_10_perf
    
```

```python
df_stat_last_10_perf.to_csv('../dataset/team_stat_last_10.csv', index=False)
```
