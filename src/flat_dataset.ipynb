```python
import pandas as pd 
```

```python
data_type = {"gameId": str, "teamId": str}
df_stat_last_10 = pd.read_csv('../dataset/team_stat_last_10.csv', dtype=data_type )
df_stat_last_10
```

```python
df_stat_last_10['col'] = df_stat_last_10.groupby('gameId').cumcount() + 1
df_stat_flatten = df_stat_last_10.pivot_table(index='gameId', columns=df_stat_last_10.groupby('gameId').cumcount() + 1, values=['teamId', 'GAME_DATE', 'DUEL', 'WL', 'NB_WIN',
       'PCT_TIR_REUSSI_L10', 'PCT_3PTS_L10', 'PCT_LANCER_FRANC_L10',
       'estimatedOffensiveRating_L10', 'offensiveRating_L10',
       'estimatedDefensiveRating_L10', 'defensiveRating_L10',
       'estimatedNetRating_L10', 'netRating_L10', 'assistPercentage_L10',
       'assistToTurnover_L10', 'assistRatio_L10',
       'estimatedTeamTurnoverPercentage_L10', 'turnoverRatio_L10',
       'effectiveFieldGoalPercentage_L10', 'trueShootingPercentage_L10',
       'estimatedPace_L10', 'pace_L10', 'pacePer40_L10', 'PIE_L10'], aggfunc='first')

df_stat_flatten.columns = [f'col{col[1]}' for col in df_stat_flatten.columns]


df_stat_flatten = df_stat_flatten.reset_index()
#df_stat_flatten
```
