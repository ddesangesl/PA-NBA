```python
import pandas as pd 
```

```python
data_type = {"gameId": str, "teamId": str}
df_stat = pd.read_csv('../dataset/merged_team_stats.csv', dtype=data_type )
df_stat
```

```python
df_stat_last_50_perf = pd.DataFrame(columns=['gameId', 'teamId', 'GAME_DATE', 'DUEL','WL','NB_WIN_L50', 'PCT_TIR_REUSSI_L50','PCT_3PTS_L50', 
                           'PCT_LANCER_FRANC_L50','estimatedOffensiveRating_L50', 'offensiveRating_L50',
                           'estimatedDefensiveRating_L50',
                           'defensiveRating_L50','estimatedNetRating_L50', 'netRating_L50', 'assistPercentage_L50',
                           'assistToTurnover_L50', 'assistRatio_L50','estimatedTeamTurnoverPercentage_L50', 'turnoverRatio_L50',
                           'effectiveFieldGoalPercentage_L50','trueShootingPercentage_L50', 'estimatedPace_L50', 'pace_L50',
                           'pacePer40_L50', 'PIE_L50'])

for team in list(set(df_stat['teamId'].tolist())):
    team_stats = df_stat[df_stat['teamId'] == team]
    team_stats.reset_index(drop=True, inplace=True)
    index = 0
    while index < len(team_stats):
        df_last_50_games = team_stats.iloc[max(0, index - 50):index]
        df_last_50_games.reset_index(drop=True, inplace=True)
        new_row = {"gameId" :  team_stats['gameId'][index],
                   "teamId" : team_stats['teamId'][index],
                   "GAME_DATE" : team_stats['GAME_DATE'][index],
                   "DUEL" : team_stats['DUEL'][index],
                   "WL" : team_stats['WL'][index],
                   "NB_WIN_L50" : (df_last_50_games['WL'] == 'W').sum(),
                   "PCT_TIR_REUSSI_L50" : df_last_50_games['PCT_TIR_REUSSI'].mean(),
                   "PCT_3PTS_L50" : df_last_50_games['PCT_3PTS'].mean(),
                   "PCT_LANCER_FRANC_L50" : df_last_50_games['PCT_LANCER_FRANC'].mean(),
                   "estimatedOffensiveRating_L50" : df_last_50_games['estimatedOffensiveRating'].mean(),
                   "offensiveRating_L50" : df_last_50_games['offensiveRating'].mean(),
                   "estimatedDefensiveRating_L50" : df_last_50_games['estimatedDefensiveRating'].mean(),
                   "defensiveRating_L50" : df_last_50_games['defensiveRating'].mean(),
                   "estimatedNetRating_L50" : df_last_50_games['estimatedNetRating'].mean(),
                   "netRating_L50" : df_last_50_games['netRating'].mean(),
                   "assistPercentage_L50" : df_last_50_games['assistPercentage'].mean(),
                   "assistToTurnover_L50" : df_last_50_games['assistToTurnover'].mean(),
                   "assistRatio_L50" : df_last_50_games['assistRatio'].mean(),
                   "estimatedTeamTurnoverPercentage_L50" : df_last_50_games['estimatedTeamTurnoverPercentage'].mean(),
                   "turnoverRatio_L50" : df_last_50_games['turnoverRatio'].mean(),
                   "effectiveFieldGoalPercentage_L50" : df_last_50_games['effectiveFieldGoalPercentage'].mean(),
                   "trueShootingPercentage_L50" : df_last_50_games['trueShootingPercentage'].mean(),
                   "estimatedPace_L50" : df_last_50_games['estimatedPace'].mean(),
                   "pace_L50" : df_last_50_games['pace'].mean(),
                   "pacePer40_L50" : df_last_50_games['pacePer40'].mean(),
                   "PIE_L50" : df_last_50_games['PIE'].mean()}
        df_stat_last_50_perf.loc[len(df_stat_last_50_perf)] = new_row
        index += 1
df_stat_last_50_perf
    
```

```python
df_stat_last_50_perf.to_csv('../dataset/team_stat_last_50.csv', index=False)
```
