```python
team_stat_flatten = team_stat_flatten.drop(columns=['DUEL', 'GAME_DATE','H_teamId', 'A_teamId'])
colonne_a_supprimer = team_stat_flatten.columns[2:]
# Boucle sur chaque paire de colonnes pour calculer la différence
for colonne in colonne_a_supprimer:
    # Extraire le préfixe de la colonne ('H_' ou 'A_')
    prefixe = colonne[0:2]
    # Extraire le nom de la colonne sans le préfixe
    nom_colonne = colonne[2:]
    # Créer une nouvelle colonne pour stocker la différence
    team_stat_flatten[f'{nom_colonne}'] = team_stat_flatten.apply(lambda row: row[f'H_{nom_colonne}'] - row[f'A_{nom_colonne}'], axis=1)

team_stat_flatten = team_stat_flatten.drop(columns=colonne_a_supprimer)
       
team_stat_flatten
```
