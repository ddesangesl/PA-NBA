```python
import pandas as pd 
```

```python
data_type = {"gameId": str, "teamId": str}
df_stat = pd.read_csv('../dataset/merged_team_stats.csv', dtype=data_type )
df_stat
```

```python
df_stat_last_30_perf = pd.DataFrame(columns=['gameId', 'teamId', 'GAME_DATE', 'DUEL','WL','NB_WIN_L30', 'PCT_TIR_REUSSI_L30','PCT_3PTS_L30', 
                           'PCT_LANCER_FRANC_L30','estimatedOffensiveRating_L30', 'offensiveRating_L30',
                           'estimatedDefensiveRating_L30',
                           'defensiveRating_L30','estimatedNetRating_L30', 'netRating_L30', 'assistPercentage_L30',
                           'assistToTurnover_L30', 'assistRatio_L30','estimatedTeamTurnoverPercentage_L30', 'turnoverRatio_L30',
                           'effectiveFieldGoalPercentage_L30','trueShootingPercentage_L30', 'estimatedPace_L30', 'pace_L30',
                           'pacePer40_L30', 'PIE_L30'])

for team in list(set(df_stat['teamId'].tolist())):
    team_stats = df_stat[df_stat['teamId'] == team]
    team_stats.reset_index(drop=True, inplace=True)
    index = 0
    while index < len(team_stats):
        df_last_30_games = team_stats.iloc[max(0, index - 30):index]
        df_last_30_games.reset_index(drop=True, inplace=True)
        new_row = {"gameId" :  team_stats['gameId'][index],
                   "teamId" : team_stats['teamId'][index],
                   "GAME_DATE" : team_stats['GAME_DATE'][index],
                   "DUEL" : team_stats['DUEL'][index],
                   "WL" : team_stats['WL'][index],
                   "NB_WIN_L30" : (df_last_30_games['WL'] == 'W').sum(),
                   "PCT_TIR_REUSSI_L30" : df_last_30_games['PCT_TIR_REUSSI'].mean(),
                   "PCT_3PTS_L30" : df_last_30_games['PCT_3PTS'].mean(),
                   "PCT_LANCER_FRANC_L30" : df_last_30_games['PCT_LANCER_FRANC'].mean(),
                   "estimatedOffensiveRating_L30" : df_last_30_games['estimatedOffensiveRating'].mean(),
                   "offensiveRating_L30" : df_last_30_games['offensiveRating'].mean(),
                   "estimatedDefensiveRating_L30" : df_last_30_games['estimatedDefensiveRating'].mean(),
                   "defensiveRating_L30" : df_last_30_games['defensiveRating'].mean(),
                   "estimatedNetRating_L30" : df_last_30_games['estimatedNetRating'].mean(),
                   "netRating_L30" : df_last_30_games['netRating'].mean(),
                   "assistPercentage_L30" : df_last_30_games['assistPercentage'].mean(),
                   "assistToTurnover_L30" : df_last_30_games['assistToTurnover'].mean(),
                   "assistRatio_L30" : df_last_30_games['assistRatio'].mean(),
                   "estimatedTeamTurnoverPercentage_L30" : df_last_30_games['estimatedTeamTurnoverPercentage'].mean(),
                   "turnoverRatio_L30" : df_last_30_games['turnoverRatio'].mean(),
                   "effectiveFieldGoalPercentage_L30" : df_last_30_games['effectiveFieldGoalPercentage'].mean(),
                   "trueShootingPercentage_L30" : df_last_30_games['trueShootingPercentage'].mean(),
                   "estimatedPace_L30" : df_last_30_games['estimatedPace'].mean(),
                   "pace_L30" : df_last_30_games['pace'].mean(),
                   "pacePer40_L30" : df_last_30_games['pacePer40'].mean(),
                   "PIE_L30" : df_last_30_games['PIE'].mean()}
        df_stat_last_30_perf.loc[len(df_stat_last_30_perf)] = new_row
        index += 1
df_stat_last_30_perf
    
```

```python
df_stat_last_30_perf.to_csv('../dataset/team_stat_last_30.csv', index=False)
```
